"""Algoritmo OrdenarMazo(mazo)
    Entrada: mazo = lista de cartas, cada carta = (palo, valor)
    Salida: mazo ordenado

    Para i desde 0 hasta longitud(mazo) - 1:
        Para j desde 0 hasta longitud(mazo) - 2 - i:
            Si comparar(mazo[j], mazo[j+1]) > 0:
                intercambiar(mazo[j], mazo[j+1])
    Retornar mazo

Función comparar(cartaA, cartaB):
    Si cartaA.palo != cartaB.palo:
        Retornar cartaA.palo - cartaB.palo
    Sino:
        Retornar cartaA.valor - cartaB.valor"""

```python
# ============================================
# ORDENAR UN MAZO DE CARTAS
# Solo con: variables, for, while, if,
# listas, tuplas, dict y funciones propias.
# ============================================

# Palos: 0=♣ Tréboles, 1=♦ Diamantes, 2=♥ Corazones, 3=♠ Picas
# Valores: 2..10, 11=J, 12=Q, 13=K, 14=A


def crear_mazo():
    mazo = []
    palo = 0
    while palo <= 3:
        valor = 2
        while valor <= 14:
            mazo = mazo + [(palo, valor)]   # concatenación, no append
            valor = valor + 1
        palo = palo + 1
    return mazo


##Valido los valores simbolicos de las cartas con letras
def nombre_valor(v):
    if v == 11:
        return "J"
    elif v == 12:
        return "Q"
    elif v == 13:
        return "K"
    elif v == 14:
        return "A"
    else:
        return str(v)


def nombre_palo(p):
    if p == 0:
        return "T"
    elif p == 1:
        return "D"
    elif p == 2:
        return "C"
    else:
        return "P"


def mostrar(mazo):
    texto = ""
    i = 0
    while i < len(mazo):
        carta = mazo[i]
        texto = texto + nombre_valor(carta[1]) + nombre_palo(carta[0])
        if i < len(mazo) - 1:
            texto = texto + ", "
        i = i + 1
    return texto


# Compara dos cartas (palo, valor).
# Devuelve: -1 si a < b, 0 si iguales, 1 si a > b
def comparar(a, b):
    if a[0] < b[0]:
        return -1
    elif a[0] > b[0]:
        return 1
    else:
        if a[1] < b[1]:
            return -1
        elif a[1] > b[1]:
            return 1
        else:
            return 0


# Ordenamiento burbuja (sin métodos)
def ordenar_mazo(mazo):
    n = len(mazo)
    i = 0
    while i < n:
        j = 0
        while j < n - 1 - i:
            if comparar(mazo[j], mazo[j + 1]) == 1:
                # intercambio manual con variable temporal
                temp = mazo[j]
                mazo[j] = mazo[j + 1]
                mazo[j + 1] = temp
            j = j + 1
        i = i + 1
    return mazo


# Mezclar el mazo (opcional, para probar).
# Usamos un "generador" simple sin import random:
def mezclar(mazo):
    n = len(mazo)
    semilla = 12345          # fija → reproducible
    i = n - 1
    while i > 0:
        # pseudoaleatorio simple (congruencial lineal)
        semilla = (semilla * 1103515245 + 12345) % 2147483648
        k = semilla % (i + 1)
        temp = mazo[i]
        mazo[i] = mazo[k]
        mazo[k] = temp
        i = i - 1
    return mazo


# ---------------- PROGRAMA PRINCIPAL ----------------

# mazo = crear_mazo()
# print("Mazo original:")
# print(mostrar(mazo))

# mezclar(mazo)
# print("\nMazo mezclado:")
# print(mostrar(mazo))
# cartas = [4D, 10D, 2T, 10C, 9P, 8P, 5T, AT, JT, 8T, KC, 3C, KT, AC, 4P, QD, 2P, JP, 6T, 8C, 7P, AD, 2D, QT, 6C, 5P, 7C, JD, 7T, QP, 9D, 10P, JC, 5D, 9T, 3P, 6D, 4C, 8D, 6P, AP, QC, 10T, 4T, 9C, 7D, 3T, 2C, 5C, KD, KP, 3D]
cartas = crear_mazo()
ordenar_mazo(cartas)
print("\nMazo ordenado:")
print(mostrar(ordenar_mazo(cartas)))
```
