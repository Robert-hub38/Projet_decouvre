```python
### Funtion récursive Factorielle
def factorielle_entier(n):
    
    if n == 0:
        return 1
    else:
        return n * factorielle_entier(n - 1)

assert factorielle_entier(0) == 1
assert factorielle_entier(1) == 1
assert factorielle_entier(3) == 6
assert factorielle_entier(4) == 24
assert factorielle_entier(5) == 120

print(factorielle_entier(0))
```

```python
### Funtion récursive Fibonacci
def fibonacci_entier(n):
    
    if n == 0:
        return 0
    elif n == 1:
        return 1
    else:
        return fibonacci_entier(n - 1) + fibonacci_entier(n - 2)

assert fibonacci_entier(8) == 21

print(fibonacci_entier(8))
```

```python
## Function récursive de la conjecture de Syracuse
def suite_syracuse(n):
    print(n, end="")
    if n == 1:
        return n
    elif n % 2 == 0:
        print(" -> ", end="")
        return suite_syracuse(n//2)
    else:
        # N = suite_syracuse(3 * n + 1)
        print("->", end="")
        return suite_syracuse(3 * n + 1)

      
suite_syracuse(15)    
```

```python
### Function récursive de la PGCD (Plus Grand Commun Diviseur)
def recursive_pgcd(a: int, b: int):
    if b == 0:
        return a
    else:
        return recursive_pgcd(b, a % b)

print(recursive_pgcd(21, 72))    
print(recursive_pgcd(12, 4))
```
