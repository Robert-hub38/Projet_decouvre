<h1>Quel Maire a planté le plus d'arbres à Grenoble </h1>

```python
##Renvoie la liste des années présentes dans le fichier
def liste_anne():
    data = data_load()
    list_anne = list()
    for x in range(len(data[1:])):
        val = data[x][19]
        if val != '' and val != 'ANNEEDEPLANTATION':
            try:
                y = int(val)
            except:
                print(val)
            list_anne.append(y)
    #list_maire = list(set(list_maire))
    #list_anne = list(set(list_anne))
    #return (sorted(list_anne))
    return list_anne


#Fonction qui calcule combien d'arbres ont été plantés par années.
#recevoir la liste des années
def plant_arbre_anne(list_anne):
    mydict = dict()
    for i in list_anne:
        if i in mydict:
            mydict[i] += 1
        else:
            mydict[i] = 1
    return(mydict)

#Retour list des maires avec nombre arbres plantés
def nombre_arbre_maire(arbres_anne):
    #maire pour periodo
    pr = [('Martin', 1949),('Michallon', 1959),('Dubedout', 1965),('Carignon', 1983),('Destot', 1995),('Piolle', 2014)]
   
    maire_periodo = dict()
   
    for i in range(len(pr)):
        x = i+1
        maire = pr[i][0]
        date_debut = pr[i][1]
        try:  
            date_fin = pr[x][1]
        except:
            err = 1
        if date_debut == date_fin:
            date_fin = 2017
        # Sommer tous les arbres plantes entre date_debut et date_fin a partir de pa
        # Assigner cette valeur a maire_periodo[maire]

        for key, val in arbres_anne.items():
            if key >= date_debut and key <= date_fin:
                if maire in maire_periodo:
                    maire_periodo[maire] += val
                else:
                    maire_periodo[maire] = val
    return(maire_periodo)

#Enregistre les arbres plante par periode
def save_periode(abres_periode):
#    import csv
    with open('../data/arbres_par_periodo.csv', 'w', newline='', encoding='utf-8') as f:
        writer = csv.writer(f)
        writer.writerow(['autor', 'nombre_arbres'])
        for autour, n in abres_periode.items():
            writer.writerow([autour,n])
        print('csv')


annees = liste_anne()
#print("===Inicio==== ")
#print("===ANNEES==== ")
#print(annees)
arbres_annees = plant_arbre_anne(annees)
#print("====ARBRES POUR ANNEES=== ")
#print(arbres_annees)
abres_periode = nombre_arbre_maire(arbres_annees)

#print("====ARBRES POUR PERIODE=== ")
print(abres_periode)


max_plant_arbre = max(abres_periode.items(), key=lambda item: item[1])
print(max_plant_arbre)

save_periode(abres_periode)
```

<h1>Histogramme nombre d'arbres plantés par année</h1>

```python
#Génère l'histogramme des arbres plantés par année.
def nombre_arbre_anne():
    lista_arbre = liste_anne_general()
    arbre_anne = dict()
    for i in lista_arbre:
        if i in arbre_anne:
            arbre_anne[i] += 1
        else:
            arbre_anne[i] = 1
    #print(arbre_anne)
    return arbre_anne

#nombre_arbre_anne()

nombre_arbre_anne = nombre_arbre_anne()

#Génère l'histogramme des arbres plantés par année.
def histogramme_arbres_anne(nombre_arbre_anne):
    annees = sorted(nombre_arbre_anne.keys())
    nomb = [nombre_arbre_anne[a] for a in annees]
    plt.figure(figsize=(14,6))
    plt.bar(annees, nomb, color='steelblue', edgecolor='black')
    plt.xlabel('Années')
    plt.ylabel('Nombres arbres plante')
    plt.title('Arbres plantés par an')
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.savefig('histogramme_arbres.png', dpi=150)
    plt.show()


histogramme_arbres_anne(nombre_arbre_anne)


    
```

<h2>Identifiez tous les genres botaniques représentés à Grenoble. Combien y en a-t-il ? </h2>

```python
#Fonction liste combien genres botanique à Grenoble, reçoit list de genres plante par annees
def combien_genres(genres):
    #lista_arbre = liste_anne_general()
    mydict = dict()
    for i in genres:
        if i in mydict:
            mydict[i] += 1
        else:
            mydict[i] = 1
    #print(genres)
    return mydict



#Fonction liste genres botaniques représentés à Grenoble
def liste_genre_botanique():
    #J'appelle la fonction qui me renvoie les données.
    data = data_load()
    data = data[1:]
    genres = []
    for x in range(len(data)):
        val = data[x][8]
        if val != '':
            try:
                y = val
            except:
                print(val)
            genres.append(y)
            genres = list(genres)     
    return genres


#Fonction genres botaniques représentés à Grenoble, reçoit list de genres plante par annees
def genres_arbre_grenoble(genres):
    genres = sorted(list(set(genres)))
    return genres



print('Lista de arbres presente à grenoble')
print(genres_arbre_grenoble(liste_genre_botanique()))

print("\t Genres d'arbres et leur nombre:")
combien_genres(liste_genre_botanique())

#liste_genre_botanique()
```

 <h1>Les espèces différentes présentes à Grenoble</h1>

```python
#Fonction liste les espèces botaniques différentes présentés à Grenoble
def liste_especes_botanique():
    #J'appelle la fonction qui me renvoie les données.
    data = data_load()
    data = data[1:]
    especes = []
    for x in range(len(data)):
        val = data[x][13]+' '+data[x][14]
        if val != ' ':
            try:
                y = val
            except:
                print(val)
            especes.append(y)
            especes = list(especes)     
    #print(especes)
    return especes
print('Les espèces différentes présentes à Grenoble')
print(genres_arbre_grenoble(liste_especes_botanique()))
```
